<font size=4 color='#002883'> **9. Verificar se existem filas com datas do track duplicadas. Em caso de existir datas duplicadas, a seguinte célula eliminas essas filas e sobrescreve (gerando) o novo arquivo track corrigido.** <font>

```python
# @title #### <font color='#002883'> Verificação de datas repetidas para evitar erro na execução do ATMOSBUD <font>
#%%capture

# Importar libraria para leitura de dataframe
import pandas as pd

# Verificação de arquivo track copiado na pasta ATMOS-BUD/inputs
track = pd.read_csv(
    TRACK_FILE_ATMOSBUD_INPUTS,
    sep=";"
)

track["time"] = pd.to_datetime(
    track["time"],
    format="%Y-%m-%d-%H%M"
)

print("=" * 60)
print("TRACK QC")
print("=" * 60)

print(f"Number of rows (original)    : {len(track)}")
print(f"Unique timestamps (original) : {track['time'].nunique()}")

duplicates = track[track["time"].duplicated(keep=False)]

if len(duplicates) > 0:
    print("\nDuplicated timestamps found:")
    print(duplicates)

    track = track.drop_duplicates(subset="time", keep="first").copy()

    print("\nDuplicated rows were removed automatically.")
else:
    print("\nNo duplicated timestamps found.")

print(f"\nNumber of rows (cleaned)    : {len(track)}")
print(f"Unique timestamps (cleaned) : {track['time'].nunique()}")

# Se quiser salvar no mesmo formato do track original:
track_to_save = track.copy()
track_to_save["time"] = track_to_save["time"].dt.strftime("%Y-%m-%d-%H%M")

track_to_save.to_csv(
    TRACK_FILE_ATMOSBUD_INPUTS,
    sep=";",
    index=False
)

print(f"\nOriginal track file overwritten: {TRACK_FILE_ATMOSBUD_INPUTS}")
```
